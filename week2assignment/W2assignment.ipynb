# Student Attendance Eligibility Checker

This program checks whether a student is eligible to move to the next session based on:

- Attendance percentage
- Number of late classes

### University Rules:
- The student must have **at least 80% attendance**.
- The university allows students to be late for **up to 2 classes**.
- The current session contains **16 classes**.

## Problem Understanding

The student has completed 16 classes in the current session.

The program will ask:

> "How many classes did you attend out of 16?"

Then it will:

1. Calculate the attendance percentage.
2. Calculate the number of classes the student did not attend.
3. Treat those missed classes as late classes.
4. Check whether the attendance is at least 80%.
5. Check whether the number of late classes is not more than 2.
6. Display whether the student is eligible to move to the next session.

### Formula:

**Attendance Percentage = (Classes Attended / Total Classes) × 100**

**Late Classes = Total Classes − Classes Attended**

```python
# Total number of classes completed in the current session
total_classes = 16
```

## Taking Student Input

The program asks the student how many classes they attended out of 16.

The `input()` function is used to take the value from the student.

Since `input()` returns a string, we convert the input into an integer using `int()`.

```python
# Ask the student how many classes they attended
attended_classes = int(input("How many classes did you attend out of 16? "))

# Display the number entered by the student
print("Classes attended:", attended_classes)
```

## Calculating Attendance Percentage

We calculate the attendance percentage using:

Attendance Percentage = (Classes Attended / Total Classes) × 100

For example, if a student attended 14 out of 16 classes:

(14 / 16) × 100 = 87.5%

Therefore, the student's attendance is 87.5%.

```python
# Calculate the attendance percentage
attendance_percentage = (attended_classes / total_classes) * 100

# Calculate the number of late/missed classes
late_classes = total_classes - attended_classes

# Display the results
print("Attendance percentage:", attendance_percentage, "%")
print("Number of late classes:", late_classes)
```

## Checking Eligibility

According to the university rules:

- Attendance must be **at least 80%**.
- The student can have **up to 2 late classes**.

Therefore, both conditions must be true:

1. Attendance percentage >= 80
2. Late classes <= 2

We use the `and` operator because both conditions must be satisfied.

```python
# Check whether the student meets the attendance requirement
attendance_requirement = attendance_percentage >= 80

# Check whether the student is within the allowed number of late classes
late_class_requirement = late_classes <= 2

# Both conditions must be satisfied
eligible = attendance_requirement and late_class_requirement

# Display the final result
if eligible:
    print("Eligible to move to the next session.")
else:
    print("Not eligible to move to the next session.")
```

## Complete Program

The following cell combines all the steps into one program.

```python
# Total number of classes completed
total_classes = 16

# Ask the student for the number of classes attended
attended_classes = int(input("How many classes did you attend out of 16? "))

# Calculate attendance percentage
attendance_percentage = (attended_classes / total_classes) * 100

# Calculate the number of late/missed classes
late_classes = total_classes - attended_classes

# Display attendance information
print("\n--- Attendance Details ---")
print("Total classes:", total_classes)
print("Classes attended:", attended_classes)
print("Attendance percentage:", attendance_percentage, "%")
print("Number of late classes:", late_classes)

# Check eligibility
if attendance_percentage >= 80 and late_classes <= 2:
    print("\nEligible to move to the next session.")
else:
    print("\nNot eligible to move to the next session.")
```

# Tuition & Scholarship Calculator

This program calculates a student's final MBA tuition based on:

- GPA and academic scholarship
- New student bonus
- Payment method
- Early enrollment

### Rules:
- GPA ≥ 3.75 → 20% scholarship
- GPA ≥ 3.50 → 15% scholarship
- GPA ≥ 3.25 → 10% scholarship
- GPA ≥ 3.00 → 5% scholarship
- GPA < 3.00 → 0% scholarship
- New student with GPA ≥ 3.50 → additional $300 scholarship
- Full payment → additional 5% discount
- Early enrollment → additional 3% discount
- Total percentage discount cannot exceed 30%

## Student Information

The program asks for:

- Original tuition
- GPA
- Student type
- Payment method
- Whether the student enrolled early

```python
# Get student information
tuition = float(input("Enter the original tuition: $"))
gpa = float(input("Enter GPA: "))

student_type = input("Enter student type (New/Continuing): ")
payment_method = input("Enter payment method (Full/Installment): ")
early_enrollment = input("Early enrollment? (Yes/No): ")
```

## Academic Scholarship

The scholarship percentage is determined by the student's GPA.

```python
# Determine academic scholarship based on GPA
if gpa >= 3.75:
    scholarship_rate = 20
elif gpa >= 3.50:
    scholarship_rate = 15
elif gpa >= 3.25:
    scholarship_rate = 10
elif gpa >= 3.00:
    scholarship_rate = 5
else:
    scholarship_rate = 0

# Calculate scholarship amount
academic_scholarship = tuition * scholarship_rate / 100
```

## Other Discounts

A new student with a GPA of 3.50 or higher receives a $300 bonus.

Full payment gives a 5% discount, while early enrollment gives a 3% discount.
The total percentage discount cannot exceed 30%.

```python
# New student bonus
if student_type == "New" and gpa >= 3.50:
    new_student_bonus = 300
else:
    new_student_bonus = 0

# Payment discount
if payment_method == "Full":
    payment_discount_rate = 5
else:
    payment_discount_rate = 0

# Early enrollment discount
if early_enrollment == "Yes":
    early_discount_rate = 3
else:
    early_discount_rate = 0

# Total percentage discount
total_discount_rate = scholarship_rate + payment_discount_rate + early_discount_rate

# Limit the total percentage discount to 30%
if total_discount_rate > 30:
    total_discount_rate = 30
```

## Final Tuition

The percentage discounts are applied to the original tuition.

The $300 new student bonus is then subtracted to find the final tuition.

```python
# Calculate total percentage discount
total_discount = tuition * total_discount_rate / 100

# Calculate final tuition after percentage discounts and $300 bonus
final_tuition = tuition - total_discount - new_student_bonus

# Prevent negative tuition
if final_tuition < 0:
    final_tuition = 0

# Display results
print("\n--- Tuition Details ---")
print("Academic Scholarship:", scholarship_rate, "%")
print("Academic Scholarship Amount: $", academic_scholarship)
print("New Student Bonus: $", new_student_bonus)
print("Payment Discount:", payment_discount_rate, "%")
print("Early Enrollment Discount:", early_discount_rate, "%")
print("Total Percentage Discount:", total_discount_rate, "%")
print("Final Tuition: $", round(final_tuition, 2))
```

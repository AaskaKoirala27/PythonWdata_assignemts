# Module 2 — Find the problems before fixing them

## 1. Count missing values

Before cleaning the dataset, we first measure how many missing values
are present in each column.

The `isna()` function identifies missing cells as `True`, while `sum()`
counts those `True` values for each column. This gives us a missing-value
profile of the dataset.

```python
missing = df.isna().sum()
print(missing)
```

## 2. Show only columns that contain missing values

The condition `missing > 0` keeps only columns that contain at least one
missing value. `sort_values()` then sorts the results from the largest
number of missing values to the smallest.

This makes it easier to identify the columns with the most missing data.

```python
missing[missing > 0].sort_values(ascending=False)
```

## 3. Calculate missing-value percentages

The missing-value percentage is calculated by dividing the number of
missing values in each column by the total number of rows and multiplying
the result by 100.

Percentages make it easier to compare missingness between columns.

```python
missing_pct = (df.isna().sum() / len(df)) * 100
missing_pct.sort_values(ascending=False)
```

## 4. Find complete duplicate rows

A duplicate row is a row where all column values are identical to a
previous row.

The `duplicated()` function identifies duplicate rows, and `sum()` counts
how many duplicate rows exist.

```python
duplicate_rows = df.duplicated().sum()
print("Duplicate rows:", duplicate_rows)
```

## 5. Check for a unique respondent identifier

The original student-performance dataset contains a `StudentID` column,
which can be checked for repeated identifiers.

The Mental Health in Tech Survey dataset does not contain a unique
respondent ID column. Therefore, a repeated-ID check is not applicable
to this dataset.

We instead use the complete duplicate-row check from the previous step
to identify identical records.

## 6. Review duplicate records

Because the dataset does not contain a unique respondent identifier and
the complete duplicate-row check returned zero duplicates, there are no
duplicate records to display or remove at this stage.

## 7. Find impossible age values

The age column contains numerical values, so we can check for values
below zero.

A negative age cannot represent a valid person's age, so any such value
would require further investigation before analysis.

```python
bad_age = df[df["What is your age?"] < 0]

bad_age[["What is your age?"]]
```

## 8. Check for suspicious age values

In addition to checking for negative ages, we can inspect the minimum and
maximum age values to identify values that may require further attention.

A value being unusual does not automatically mean that it is incorrect,
so suspicious values should be investigated rather than automatically
deleted.

```python
print("Minimum age:", df["What is your age?"].min())
print("Maximum age:", df["What is your age?"].max())
```

## 9. Check important numerical ranges

A dictionary can store sensible lower and upper limits for numerical
variables.

The loop applies the same validation pattern to each variable.

For age, values below 0 are considered invalid. Missing values are
excluded from the invalid-value count using `notna()`.

```python
checks = {
    "What is your age?": (0, 120)
}

for col, (low, high) in checks.items():
    bad = df[col].notna() & ((df[col] < low) | (df[col] > high))
    print(col, ":", bad.sum(), "invalid values")
```

# Module 3 — Clean the Dataset

## 1. Create a copy before cleaning

We create a separate copy of the original DataFrame so that the original
data remains unchanged.

All cleaning operations will be performed on the new `clean` DataFrame.

```python
clean = df.copy()
```

## 2. Remove duplicate rows

We use `drop_duplicates()` to remove completely duplicated records.

The dataset contains no duplicate rows, so this operation should not
change the number of records.

```python
clean = clean.drop_duplicates()
```

```python
print("Rows after removing duplicates:", len(clean))
```

## 3. Replace impossible age values with missing values

The age column contains values that are outside a reasonable range.

Instead of deleting the entire respondent's row, we replace the invalid
age value with a missing value.

This allows us to preserve the rest of the respondent's information.

```python
age_col = "What is your age?"

clean.loc[
    (clean[age_col] < 0) | (clean[age_col] > 120),
    age_col
] = pd.NA
```

## 4. Fill missing numerical values

For numerical columns, we use the median value to replace missing
values.

The median is the middle value when the observations are ordered and is
less affected by extreme values than the mean.

```python
numeric_columns = clean.select_dtypes(include="number").columns

for col in numeric_columns:
    if clean[col].isna().any():
        clean[col] = clean[col].fillna(clean[col].median())
```

## 5. Handle missing binary values

Some numerical columns in the survey are actually binary variables,
where 0 and 1 represent two possible responses.

Following the simple classroom cleaning rule, missing values in these
binary columns are replaced with 0.

```python
binary_columns = [
    "Is your employer primarily a tech company/organization?",
    "Is your primary role within your company related to tech/IT?",
    "Do you have medical coverage (private insurance or state-provided) which includes treatment of \xa0mental health issues?"
]

for col in binary_columns:
    clean[col] = clean[col].fillna(0)
```

## 6. Fill missing categorical values

Categorical columns contain text-based responses rather than numerical
measurements.

For missing categorical responses, we use the explicit category
`"Unknown"` instead of inventing a response such as `"Yes"` or `"No"`.

This makes the missing response visible in later analysis.

```python
categorical_columns = clean.select_dtypes(include="str").columns

for col in categorical_columns:
    clean[col] = clean[col].fillna("Unknown")
```

## 7. Recheck missing values

After cleaning, we check the dataset again to make sure that no missing
values remain.

This is an important verification step because cleaning code should
always be checked rather than assumed to have worked.

```python
remaining_missing = clean.isna().sum()

print(remaining_missing[remaining_missing > 0])
```

```python
print("Duplicate rows after cleaning:", clean.duplicated().sum())
```

## 9. Recheck invalid age values

We repeat the age-range check after cleaning to confirm that no ages
below 0 or above 120 remain.

```python
bad_age = clean[
    (clean["What is your age?"] < 0) |
    (clean["What is your age?"] > 120)
]

print("Invalid age values:", len(bad_age))
```

```python
#Final verification
print("Rows:", clean.shape[0])
print("Columns:", clean.shape[1])
print("Missing values:", clean.isna().sum().sum())
print("Duplicate rows:", clean.duplicated().sum())
```

# Module 4 — Ask Questions and Analyze the Data

## 1. What is the average age of respondents?

We calculate the mean age of respondents using the cleaned dataset.

The mean provides a simple summary of the typical age represented in
the survey.

```python
average_age = clean["What is your age?"].mean()

print(f"Average age: {average_age:.2f}")
```

## 2. Find the oldest respondents

We sort the cleaned dataset by age in descending order.

This allows us to inspect the respondents with the highest recorded ages.

```python
oldest = clean.sort_values(
    by="What is your age?",
    ascending=False
)

oldest[["What is your age?", "What is your gender?"]].head(10)
```

## 3. How many respondents are in each gender category?

We use `value_counts()` to count how many respondents fall into each
gender category.

The `dropna=False` option keeps missing responses visible. Since we
cleaned missing categorical values as `"Unknown"`, this also allows us
to see the `"Unknown"` category if it exists.

```python
gender_counts = clean["What is your gender?"].value_counts(
    dropna=False
)

print(gender_counts)
```

## 4. Does having mental-health benefits relate to seeking treatment?

This is our adaptation of the teacher's:

`Parental support vs Final Grade`


Our dataset contains variables related to workplace mental-health support and treatment.

We can compare:

Does your employer provide mental health benefits?

with:

Have you sought treatment for a mental health condition?

```python
print(clean["Does your employer provide mental health benefits as part of healthcare coverage?"].value_counts())
```

```python
print(clean["Have you ever sought treatment for a mental health issue from a mental health professional?"].value_counts())
   
```

```python
treatment_by_benefits = (
    clean.groupby(
        "Does your employer provide mental health benefits as part of healthcare coverage?"
    )["Have you ever sought treatment for a mental health issue from a mental health professional?"]
    .mean()
)

print(treatment_by_benefits)
```

## 5. Compare mental-health treatment responses across age groups

We divide respondents into age bands using `pd.cut()` and then compare
the average treatment response across those groups.

This demonstrates how a numerical variable can be converted into useful
categories for group analysis.

```python
clean["AgeGroup"] = pd.cut(
    clean["What is your age?"],
    bins=[0, 20, 30, 40, 50, 60, 120],
    labels=[
        "Under 21",
        "21-30",
        "31-40",
        "41-50",
        "51-60",
        "61+"
    ]
)
```

```python
treatment_by_age = (
    clean.groupby("AgeGroup", observed=True)[
        "Have you ever sought treatment for a mental health issue from a mental health professional?"
    ]
    .mean()
)

print(treatment_by_age)
```

## 6. Filter respondents who reported a mental health disorder

We use Boolean filtering to identify respondents who reported having a
mental health disorder.

This demonstrates how conditions can be used to create a subset of the
dataset.

```python
mental_health_group = clean[
    clean["Do you currently have a mental health disorder?"] == "Yes"
]

print("Number of respondents:", len(mental_health_group))
```

```python
mental_health_group[
    [
        "What is your age?",
        "What is your gender?",
        "Do you currently have a mental health disorder?"
    ]
].head(10)
```

## 7. Examine the relationship between age and treatment

We calculate the correlation between respondent age and the numerical
treatment-response variable.

Correlation measures the strength and direction of a linear relationship
between two numerical variables.

```python
correlation = clean[
    [
        "What is your age?",
        "Have you ever sought treatment for a mental health issue from a mental health professional?"
    ]
].corr()

print(correlation)
```

# Module 5 — Visualize the Data

```python
import matplotlib.pyplot as plt
```

```python
#Histogram — Distribution of Respondent Ages
plt.figure(figsize=(8, 5))

plt.hist(
    clean["What is your age?"],
    bins=10
)

plt.xlabel("Age")
plt.ylabel("Number of Respondents")
plt.title("Distribution of Respondent Ages")

plt.show()
```

```python
#Bar Chart — Mental-Health Treatment Responses
treatment_counts = clean[
    "Have you ever sought treatment for a mental health issue from a mental health professional?"
].value_counts()

treatment_counts.index = treatment_counts.index.map({
    0: "No",
    1: "Yes"
})

plt.figure(figsize=(8, 5))

plt.bar(
    treatment_counts.index,
    treatment_counts.values
)

plt.xlabel("Sought Professional Treatment")
plt.ylabel("Number of Respondents")
plt.title("Respondents Who Have Sought Mental-Health Treatment")

plt.show()
```

```python
summary["Treatment Rate"] = summary["Treatment Rate"] * 100

print(summary)
```
